```python
# eCommerce and logistic project Preprocessing code

import pandas as pd
from pathlib import Path

# loading the csv file for preprocessing

df = pd.read_csv("BADS_WS1920_known.csv")
df=pd.DataFrame(df)

print(df)
```

```python
df.head()
```

```python
# Display total rows and columns

print(df.shape)
```

```python
#Inspect if all required columns are present

required_columns = [
    "order_item_id",
    "order_date",
    "delivery_date",
    "item_id",
    "item_size",
    "item_color",
    "brand_id",
    "item_price",
    "user_id",
    "user_title",
    "user_dob",
    "user_state",
    "user_reg_date",
    "return"]

missing_columns = [column for column in required_columns if column not in df.columns]

if missing_columns:
    raise ValueError(f"The following required columns are missing: {missing_columns}")

print("\nAll required columns are present.")
```

```python
#Inspect dataset for any duplicated rows

duplicate_count = df.duplicated().sum()

print(f"\nNumber of duplicate rows: {duplicate_count}")

df = df.drop_duplicates()

print(f"Rows after removing duplicates: {len(df)}")
```

```python
#Inspect dataset for any duplicated columns

duplicate_columns = df.columns[df.T.duplicated()]

print("Duplicate columns:", list(duplicate_columns))

df = df.loc[:, ~df.T.duplicated()]

print("Columns after removing duplicates:", len(df.columns))
```

```python
#Converting all date columns to date objects

date_columns = [
    "order_date",
    "delivery_date",
    "user_dob",
    "user_reg_date"]

for column in date_columns:
    df[column] = pd.to_datetime(df[column],errors="coerce")

print("\nDate columns converted successfully.")
```

```python
#Display the converted date columns for inspection

date_columns = [
    "order_date",
    "delivery_date",
    "user_dob",
    "user_reg_date"]

for column in date_columns:
    df[column] = pd.to_datetime(df[column], errors="coerce")

    print(f"\n{column}:")
    print("Data type:", df[column].dtype)
    print("Sample values:")
    print(df[column].head())
    print("Values that could not be parsed:", df[column].isna().sum())
    
```

```python
#Cleaning categorical columns into clean strings

categorical_columns = [
    "item_size",
    "item_color",
    "user_title",
    "user_state"]

for column in categorical_columns:

    df[column] = (df[column].astype("string").str.strip().str.lower())

    print(f"\n{column}")
    print("Data type:", df[column].dtype)
    print("Sample values:")
    print(df[column].head())
    print("Unique values:", df[column].nunique())
```

```python
df["item_color"] = df["item_color"].replace(
    {"?": pd.NA, "": pd.NA})

blank_count = df["item_color"].isna().sum()
blank_percent = blank_count / len(df) * 100 if len(df) else 0

print(f"Blank/missing item_color values: {blank_count}")
print(f"Percentage missing: {blank_percent:.2f}%")

if blank_percent < 10 and blank_count > 0:
    mode = df["item_color"].mode(dropna=True)

    if not mode.empty:
        df["item_color"] = df["item_color"].fillna(mode.iloc[0])
        print(f"Filled missing values with the most frequent color: {mode.iloc[0]!r}")
    else:
        print("Cannot impute: item_color has no non-missing values.")
else:
    print("Missing values were not imputed (percentage is 10% or higher, or none are missing).")
```

```python
#Convert columns to numeric; unconvertible values become NaN

numeric_columns = [
    "order_item_id",
    "item_id",
    "brand_id",
    "item_price",
    "user_id"]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Handle user_id separately

user_id_missing = df["user_id"].isna().sum()
user_id_missing_pct = user_id_missing / len(df) * 100 if len(df) else 0

print(f"Missing user_id values: {user_id_missing} ({user_id_missing_pct:.2f}%)")

#Keep User IDs as strings and call the missing values "anonymous".

if user_id_missing_pct < 1:

    df["user_id"] = df["user_id"].astype("string").fillna("anonymous")
    print("Filled missing user_id values with 'anonymous'.")
else:
    df = df.dropna(subset=["user_id"])
    print("Dropped rows with missing user_id values.")

#Impute other numeric columns with their mean if less than 10% are missing

for column in numeric_columns:
    if column == "user_id":
        continue

    missing_count = df[column].isna().sum()
    missing_pct = missing_count / len(df) * 100 if len(df) else 0

    print(f"{column}: {missing_count} missing ({missing_pct:.2f}%)")

    if 0 < missing_pct < 10:
        mean_value = df[column].mean()
        if pd.notna(mean_value):
            df[column] = df[column].fillna(mean_value)
            print(f"  Imputed with mean: {mean_value:.2f}")
    elif missing_count > 0:
        print("  Not imputed: missing percentage is 10% or higher.")
```

```python

# Convert the target column to numbers.Values that cannot be converted become NaN.

df["return"] = pd.to_numeric(df["return"], errors="coerce")

# Display missing target values

missing_count = df["return"].isna().sum()
print(f"Missing values in 'return': {missing_count}")

if missing_count > 0:
    raise ValueError("The target column 'return' contains missing values.")

# Check that the target contains only 0 and 1

unique_target_values = set(df["return"].unique())
print(f"Unique values in 'return': {sorted(unique_target_values)}")

if not unique_target_values.issubset({0, 1}):
    raise ValueError("The 'return' column must contain only 0 and 1.")

#Display counts and percentages for each class

return_counts = df["return"].value_counts().reindex([0, 1], fill_value=0)
return_percentages = df["return"].value_counts(normalize=True).reindex(
    [0, 1], fill_value=0) * 100

print("\nReturn target summary:")
print(f"Not returned (0): {return_counts[0]:,} ({return_percentages[0]:.2f}%)")
print(f"Returned (1):     {return_counts[1]:,} ({return_percentages[1]:.2f}%)")
print(f"Total records:    {len(df):,}")

print("\nValidation successful: 'return' contains no missing values and only 0/1.")
```

```python
#Display total count and percentage of returned products and not returned

counts = df["return"].value_counts()
percentages = df["return"].value_counts(normalize=True) * 100

returned = counts.get(1, 0)
kept = counts.get(0, 0)

returned_pct = percentages.get(1, 0)
kept_pct = percentages.get(0, 0)

print(f"Total items: {len(df)}")
print(f"Returned: {returned} ({returned_pct:.2f}%)")
print(f"Kept (not returned): {kept} ({kept_pct:.2f}%)")
```
